```python
from crewai_tools import WebsiteSearchTool
import os

web_search_tool = WebsiteSearchTool()
```

```python
import os
from langchain_openai import ChatOpenAI

if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("Set the OPENAI_API_KEY environment variable before running this cell.")

llm = ChatOpenAI(temperature=0, model="gpt-4o-mini")
```

```python
from crewai import Agent

researcher = Agent(
    role = "Market Research Analyst",
    goal = "Provide up-to-date market analysis of AI Industry",
    backstory = "An expert analyst with keen eye for market trends",
    tools = [web_search_tool],
    verbose = True
)
```

```python
writer = Agent(
    role = "Content Writer",
    goal = "Craft engaging blog posts about AI industry",
    backstory = "A skilled writer with a passion for technology",
    tools = [],
    verbose = True
)
```

```python
from crewai import Task, Crew

research = Task(
    description = "Search on web for latest AI trends and provide a summarized report",
    expected_output = "A summary of top 3 trending developments in AI with insights on their impact",
    agent = researcher
)
```

```python
write = Task(
    description = "Write an engaging blog post about AI industry based research analyst's summary",
    expected_output = "A well-structured, 4-paragraph log post in markdown format with simple, engaging content",
    agent = writer,
    output_file= 'blog-posts/news.md'
)
```

```python
crew = Crew(
    agents = [researcher, writer],
    tasks = [research, write],
    verbose = True,
    planning = True
)

await crew.kickoff_async()
```
